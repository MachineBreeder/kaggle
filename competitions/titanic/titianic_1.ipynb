- **"0"** → 해당 승객은 **사망**

제출 파일은 다음과 같은 형식이어야 합니다:  
- **"PassengerId"** 열: **test.csv**에 있는 각 승객의 ID를 포함해야 합니다.  
- **"Survived"** 열: 여러분이 예측한 생존 여부를 포함해야 합니다.  
  - **"1"** → 승객이 생존했다고 예측  
  - **"0"** → 승객이 사망했다고 예측  

<span style="font-size: 12px;"> The **gender_submission.csv** file is provided as an example that shows how you should structure your predictions. But, just like this file, your submission should have:
- <span style="font-size: 12px;"> a **"PassengerId"** column containing the IDs of each passenger from **test.csv**.
- <span style="font-size: 12px;"> a **"Survived"** column (that you will create!) with a "1" for the rows where you think the passenger survived, and a "0" where you predict that the passenger died.

이 가정이 합리적인지 확인하기 위해, **train.csv에서 실제 데이터와 비교해 보겠습니다.**

위 코드는 **train.csv에서 남성 승객이 생존한 비율을 계산**합니다.  

결과를 보면,

따라서, **성별이 생존 여부를 예측하는 중요한 요소임을 알 수 있습니다**.  
이러한 이유로, **gender_submission.csv** 파일의 성별 기반 예측은 **나쁘지 않은 첫 번째 가정**이라고 할 수 있습니다.  

하지만 이 가정은 **오직 하나의 열(성별)만을 고려한 예측**입니다.

- 하지만 여러 열을 동시에 분석하는 것은 **매우 어렵고 시간이 많이 걸립니다**.  

따라서, **머신 러닝을 활용하여 이 과정을 자동화하고 더 나은 예측을 수행하겠습니다.**

<span style="font-size: 12px;"> But at the end of the day, this gender-based submission bases its predictions on only a single column. Since it is quite difficult to consider several columns at once (or, it would take a long time to consider all possible patterns in many different columns simultaneously), we'll use machine learning to automate this for us.

## Your first machine learning model

우리는 **랜덤 포레스트(Random Forest) 모델**을 구축할 것입니다.  

랜덤 포레스트는 여러 개의 **"트리(Tree)"**로 구성된 모델입니다.  
- 각 트리는 개별 승객 데이터를 바탕으로 **생존 여부를 예측**합니다.  
- 이후, **모든 트리의 투표 결과를 종합하여 최종 결정을 내립니다**.  
- (_아래 그림에서는 3개의 트리가 있지만, 우리는 **100개의 트리를 생성**할 것입니다._)  

![](https://i.imgur.com/AC9Bq63.png)  

아래 코드에서는 **train.csv**의 네 개의 열 (**"Pclass"**, **"Sex"**, **"SibSp"**, **"Parch"**)을 활용하여 랜덤 포레스트 모델을 훈련합니다.  
- 학습이 끝나면, **test.csv에 대한 예측을 수행**합니다.  
- 예측 결과는 CSV 파일 **submission.csv**로 저장됩니다.  

이제 아래 코드를 **새로운 코드 셀**에 입력한 후 실행하세요.

<span style="font-size: 12px;"> We'll build what's known as a **random forest model**.  This model is constructed of several "trees" (there are three trees in the picture below, but we'll construct 100!) that will individually consider each passenger's data and vote on whether the individual survived.  Then, the random forest model makes a democratic decision: the outcome with the most votes wins!

<span style="font-size: 12px;"> The code cell below looks for patterns in four different columns (**"Pclass"**, **"Sex"**, **"SibSp"**, and **"Parch"**) of the data.  It constructs the trees in the random forest model based on patterns in the **train.csv** file, before generating predictions for the passengers in **test.csv**.  The code also saves these new predictions in a CSV file **submission.csv**.

<span style="font-size: 12px;"> Copy this code into your notebook, and run it in a new code cell.

```python
from sklearn.ensemble import RandomForestClassifier

y = train_data["Survived"]

features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)
predictions = model.predict(X_test)

output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('titanic_submission.csv', index=False)
print("Your submission was successfully saved!")
```
