### 모델링을 위한 데이터 선택

귀하의 데이터 세트에는 이해하기 힘들거나, 심지어 멋지게 인쇄하기 어려울 만큼 변수가 너무 많습니다. 이 엄청난 양의 데이터를 이해할 수 있는 수준으로 어떻게 줄일 수 있을까요?

우리는 직감을 사용하여 몇 가지 변수를 선택하는 것으로 시작할 것입니다.<br>
이후 과정에서는 변수의 우선순위를 자동으로 지정하는 통계적 기술을 보여드리겠습니다.

변수/열을 선택하려면 데이터세트의 모든 열 목록을 확인해야 합니다.<br>
이는 DataFrame의 열 속성(아래 코드의 맨 아래 줄)으로 수행됩니다.

```python
import pandas as pd

melbourne_file_path = '../input/melbourne-housing-snapshot/melb_data.csv'
melbourne_data = pd.read_csv(melbourne_file_path) 
melbourne_data.columns
```

```python
# The Melbourne data has some missing values (some houses for which some variables weren't recorded.)
# We'll learn to handle missing values in a later tutorial.  
# Your Iowa data doesn't have missing values in the columns you use. 
# So we will take the simplest option for now, and drop houses from our data. 
# Don't worry about this much for now, though the code is:

# dropna drops missing values (think of na as "not available")
melbourne_data = melbourne_data.dropna(axis=0)
```

데이터 하위 집합을 선택하는 방법은 여러 가지가 있습니다.<br>
Pandas 과정에서는 이를 더 자세히 다루지만 지금은 두 가지 접근 방식에 집중하겠습니다.

* "예측 대상"을 선택하는 데 사용하는 점 표기법
* "특징"을 선택하는 데 사용하는 열 목록을 사용하여 선택

### 예측 대상 선택
점 표기법 으로 변수를 꺼낼 수 있습니다. <br>
이 단일 열은 Series 에 저장되며 , 이는 단일 데이터 열만 있는 DataFrame과 크게 비슷합니다.

점 표기법을 사용하여 예측하고자 하는 열을 선택합니다.<br>
이를 예측 대상 이라고 합니다.<br>
관례에 따라 예측 대상은 y 라고 합니다.<br>
따라서 멜버른 데이터의 주택 가격을 저장하는 데 필요한 코드는 다음과 같습니다.

```python
y = melbourne_data.Price
```

### "기능" 선택하기 
모델에 입력된 열(그리고 나중에 예측을 위해 사용되는 열)을 "특징"이라고 합니다.<br>
우리의 경우, 이는 주택 가격을 결정하는 데 사용되는 열입니다.<br>
때로는 타겟을 제외한 모든 열을 특징으로 사용할 것입니다.<br>
다른 때는 더 적은 특징을 사용하는 것이 더 나을 것입니다.

지금은 몇 가지 기능만 있는 모델을 빌드합니다.<br>
나중에는 다른 기능으로 빌드된 모델을 반복하고 비교하는 방법을 살펴보겠습니다.

우리는 괄호 안에 열 이름 목록을 제공하여 여러 기능을 선택합니다.<br>
그 목록의 각 항목은 문자열(따옴표 포함)이어야 합니다.

다음은 그 예입니다.

```python
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']
```

관례에 따라 이 데이터를 X 라고 합니다.

```python
X = melbourne_data[melbourne_features]
```

주택가격을 예측하는데 사용할 describe방법과 head상위 몇 행을 보여주는 방법을 간략히 살펴보겠습니다.

### 모델 구축 
scikit-learn 라이브러리를 사용하여 모델을 만듭니다.<br>
코딩할 때 이 라이브러리는 샘플 코드에서 볼 수 있듯이 sklearn 으로 작성됩니다. <br>
Scikit-learn은 일반적으로 DataFrames에 저장된 데이터 유형을 모델링하는 데 가장 인기 있는 라이브러리입니다.

모델을 구축하고 사용하는 단계는 다음과 같습니다.

* 정의: 어떤 유형의 모델일까요? 의사결정 트리? 다른 유형의 모델? 모델 유형의 다른 매개변수도 지정됩니다.
* 적합성: 제공된 데이터에서 패턴을 포착합니다. 이것이 모델링의 핵심입니다.
* 예측: 들리는 그대로입니다
* 평가 : 모델의 예측이 얼마나 정확한지 확인합니다.

다음은 scikit-learn을 사용하여 의사결정 트리 모델을 정의하고 이에 특성과 대상 변수를 적용하는 예입니다.

```python
from sklearn.tree import DecisionTreeRegressor

# Define model. Specify a number for random_state to ensure same results each run
melbourne_model = DecisionTreeRegressor(random_state=1)

# Fit model
melbourne_model.fit(X, y)
```

많은 머신 러닝 모델은 모델 학습에서 어느 정도의 무작위성을 허용합니다. <br>
숫자를 지정하면 random_state각 실행에서 동일한 결과를 얻을 수 있습니다. <br>
이는 좋은 관행으로 간주됩니다. <br>
어떤 숫자를 사용하든 모델 품질은 선택한 값에 따라 크게 달라지지 않습니다. <br>

이제 우리는 예측을 위해 사용할 수 있는 적합 모델을 가지고 있습니다.

실제로는 이미 가격이 매겨진 주택이 아니라 시장에 새로 나올 주택에 대한 예측을 하고 싶을 것입니다.<br>
하지만 예측 함수가 어떻게 작동하는지 보기 위해 훈련 데이터의 처음 몇 행에 대한 예측을 하겠습니다.

```python
print("Making predictions for the following 5 houses:")
print(X.head())
print("The predictions are")
print(melbourne_model.predict(X.head()))
```
