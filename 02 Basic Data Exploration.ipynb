Pandas를 사용하여 데이터에 익숙해지기 ¶
머신 러닝 프로젝트의 첫 단계는 데이터에 익숙해지는 것입니다. 이를 위해 Pandas 라이브러리를 사용합니다. Pandas는 데이터 과학자들이 데이터를 탐색하고 조작하는 데 사용하는 주요 도구입니다. 대부분의 사람들은 코드에서 pandas를 pd.로 줄여서 사용합니다. 다음 명령어를 사용하면 됩니다 .

```python
import pandas as pd
```

Pandas 라이브러리에서 가장 중요한 부분은 DataFrame입니다. DataFrame은 테이블과 같은 유형의 데이터를 저장합니다. Excel의 시트나 SQL 데이터베이스의 테이블과 유사합니다.

Pandas는 이러한 유형의 데이터로 하고 싶은 대부분의 작업에 대해 강력한 방법을 제공합니다.

예를 들어 호주 멜버른의 주택 가격 데이터를 살펴보겠습니다 . 실습에서는 아이오와주의 주택 가격이 포함된 새로운 데이터세트에 동일한 과정을 적용해 보겠습니다.

예제(멜버른) 데이터는 파일 경로에 있습니다 ../input/melbourne-housing-snapshot/melb_data.csv.

다음 명령을 사용하여 데이터를 로드하고 탐색합니다.

```python
# save filepath to variable for easier access
melbourne_file_path = '../input/melbourne-housing-snapshot/melb_data.csv'

# read the data and store data in DataFrame titled melbourne_data
melbourne_data = pd.read_csv(melbourne_file_path) 

# print a summary of the data in Melbourne data
melbourne_data.describe()
```

데이터 설명 해석 ¶
결과는 원래 데이터세트의 각 열에 대해 8개의 숫자를 보여줍니다. 첫 번째 숫자인 count 는 결측값이 없는 행의 수를 나타냅니다.

결측값은 여러 가지 이유로 발생합니다. 예를 들어, 침실 1개짜리 주택을 조사할 때 두 번째 침실의 크기는 수집되지 않습니다. 결측값에 대해서는 나중에 다시 다루겠습니다.

두 번째 값은 평균 (mean) 입니다 . 그 아래에 있는 std 는 표준편차(standard deviation)로, 값들이 수치적으로 얼마나 퍼져 있는지를 측정합니다.

최소값 , 25% , 50% , 75% , 최대 값을 해석하려면 각 열을 가장 낮은 값에서 가장 높은 값 순으로 정렬한다고 가정해 보겠습니다. 첫 번째(가장 작은) 값은 최소값입니다. 목록을 4분의 1 정도 살펴보면 값의 25%보다 크고 75%보다 작은 값을 찾을 수 있습니다. 이것이 25% 값("25th percentile"이라고 발음)입니다. 50번째 백분위수와 75번째 백분위수도 유사하게 정의되며, 최대값 은 가장 큰 값입니다.
