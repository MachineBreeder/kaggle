### 소개 

<span style="font-size: 15px;"> 머신 러닝 모델의 작동 방식과 사용 방법에 대한 개요부터 살펴보겠습니다. <br>
<span style="font-size: 15px;"> 통계 모델링이나 머신 러닝을 해 본 경험이 있다면 기본적인 내용일 수 있습니다.

##### 이 과정에서는 다음 시나리오를 진행하면서 모델을 구축하게 됩니다.

<span style="font-size: 15px;"> 사촌이 부동산 투자로 수백만 달러를 벌었습니다. <br>
데이터 과학에 대한 당신의 관심 때문에 사업 파트너가 되겠다고 제안했습니다. <br>
사촌이 돈을 대주고, 당신은 다양한 주택의 가치를 예측하는 모델을 제시하면 됩니다.

<span style="font-size: 15px;"> 사촌에게 과거에 부동산 가격을 어떻게 예측했는지 물어보니, 그는 그저 직감이라고 대답합니다. <br>
하지만 더 자세히 물어보니, 과거에 본 집들의 가격 패턴을 파악하고, 그 패턴을 바탕으로 현재 고려 중인 새 집을 예측한다는 것을 알게 됩니다.

<span style="font-size: 15px;"> 머신 러닝도 같은 방식으로 작동합니다. <br>
먼저 의사결정 트리라는 모델부터 살펴보겠습니다. <br>
더 정확한 예측을 제공하는 더 정교한 모델들이 있습니다. <br>
하지만 의사결정 트리는 이해하기 쉽고, 데이터 과학 분야에서 최고의 모델들을 구성하는 기본 요소입니다.

<span style="font-size: 15px;"> 단순화를 위해 가능한 가장 간단한 의사결정 트리부터 시작해 보겠습니다.

<span style="font-size: 15px;"> 이 모델은 주택을 두 가지 범주로만 구분합니다. <br>
고려 중인 모든 주택의 예상 가격은 동일 범주 내 주택의 과거 평균 가격입니다.

<span style="font-size: 15px;"> 데이터를 사용하여 주택을 두 그룹으로 나누는 방법을 결정하고, 각 그룹의 예상 가격을 다시 계산합니다. <br>
데이터에서 패턴을 포착하는 이 단계를 모델 적합(fitting) 또는 학습(training) 이라고 합니다. <br>
모델 적합에 사용되는 데이터를 학습 데이터(training data) 라고 합니다.

<span style="font-size: 15px;"> 모델 적합 방식(예: 데이터 분할 방법)에 대한 세부 사항은 매우 복잡하므로 나중에 다루도록 하겠습니다. <br>
모델 적합이 완료되면 새로운 데이터에 적용하여 추가 주택 가격을 예측할 수 있습니다.

#### 의사결정 트리 개선

<span style="font-size: 15px;"> 다음 두 가지 의사결정 트리 중 부동산 교육 데이터를 적용하여 만들어질 가능성이 더 높은 것은 무엇입니까?

##### 왼쪽의 의사결정 트리(의사결정 트리 1) 가 더 타당해 보입니다.
<span style="font-size: 15px;"> 침실이 많은 집이 침실이 적은 집보다 더 높은 가격에 매매되는 경향이 있다는 현실을 잘 반영하고 있기 때문입니다. <br>
<span style="font-size: 15px;"> 이 모델의 가장 큰 단점은 욕실 수, 대지 크기, 위치 등 주택 가격에 영향을 미치는 대부분의 요소를 제대로 반영하지 못한다는 것입니다.

<span style="font-size: 15px;"> 더 많은 "분할"을 갖는 트리를 사용하면 더 많은 요소를 포착할 수 있습니다. <br>
<span style="font-size: 15px;"> 이러한 트리를 "더 깊은" 트리라고 합니다. <br>
<span style="font-size: 15px;"> 각 주택의 전체 대지 크기까지 고려하는 의사결정 트리는 다음과 같습니다.

<span style="font-size: 15px;"> 의사결정 트리를 따라가며 주택 가격을 예측할 때, 항상 해당 주택의 특성에 맞는 경로를 선택합니다. <br>
<span style="font-size: 15px;"> 주택의 예측 가격은 트리의 맨 아래에 있습니다. 예측을 내리는 맨 아래 지점을 잎(leaf)이라고 합니다.

<span style="font-size: 15px;"> 잎에서의 분할과 값은 데이터에 따라 결정되므로, 작업할 데이터를 확인할 차례입니다.
